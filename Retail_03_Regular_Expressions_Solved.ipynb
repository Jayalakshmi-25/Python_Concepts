### Q1. Find a product code (retail variant 1).

```python
import re
text='SALE SKU001 quantity=2 price=₹101.50 order ORD-1001'
pattern=re.compile(r'\bSKU\d{3}\b',re.IGNORECASE)
print('matches:',pattern.findall(text))
print('numbers:',re.findall(r'\d+(?:\.\d+)?',text))
```

**Explanation:** Raw strings keep regex escapes readable. Anchors or `fullmatch` are used for validation, named groups improve parsing, and bounded patterns reduce accidental partial matches.

### Q31. Parse a receipt line into named groups (retail variant 1).

```python
import re
line='SKU101 | 3 | 249.50'
pat=re.compile(r'^(?P<sku>[A-Z]{3}\d{3})\s*\|\s*(?P<qty>\d+)\s*\|\s*(?P<price>\d+(?:\.\d{2})?)$')
m=pat.fullmatch(line); print(m.groupdict() if m else 'invalid')
```

**Explanation:** Raw strings keep regex escapes readable. Anchors or `fullmatch` are used for validation, named groups improve parsing, and bounded patterns reduce accidental partial matches.
