### Q1. Write a function for line total (retail variant 1).

```python
from functools import reduce
def line_total(price:float,qty:int=1)->float:
    '''Return quantity multiplied by unit price.'''
    return round(price*qty,2)
prices=[11,20,30]
print(line_total(prices[0],2))
print(list(map(lambda p:round(p*1.05,2),prices)))
print(list(filter(lambda p:p>=25,prices)))
print(reduce(lambda a,b:a+b,prices,0))
```

**Explanation:** The function has one clear responsibility. Lambdas remain short, transformations are separated from selection, and `reduce` uses an explicit initializer so empty input is safe.

### Q35. Write a decorator that logs checkout calls (retail variant 1).

```python
from functools import wraps
def audit(label):
    def deco(fn):
        @wraps(fn)
        def wrapper(*a,**k):
            print(label,fn.__name__); return fn(*a,**k)
        return wrapper
    return deco
@audit('CHECKOUT')
def total(values): return sum(values)
print(total([10,20]))
```

**Explanation:** The function has one clear responsibility. Lambdas remain short, transformations are separated from selection, and `reduce` uses an explicit initializer so empty input is safe.
